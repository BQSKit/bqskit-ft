```python
from bqskit.ir import Circuit
from bqskit.ir.gates import HGate, SqrtTGate, TGate, XGate, CNOTGate
from bqskit.ft.gates.logical_and import LogicalAndGate, LogicalAndDgGate

# Test Catylzed sqrt(T) gate synthesis
circuit = Circuit(4)
# Catalyzed sqrt(T) gate on qubit 2
circuit.append_gate(HGate(), [2])
circuit.append_gate(SqrtTGate(), [2])

# Should kick back sqrt(T) on qubit 0 and 1
circuit.append_gate(HGate(), [0])
circuit.append_gate(HGate(), [1])


# Requires a T state input on qubit 3 for LogicalAndGate
circuit.append_gate(HGate(), [3])
circuit.append_gate(TGate(), [3])

# Now run circuit

circuit.append_gate(XGate(), [2])
circuit.append_gate(CNOTGate(), [0, 1])
circuit.append_gate(CNOTGate(), [0, 2])

circuit.append_gate(LogicalAndGate(), [1, 2, 3])

circuit.append_gate(CNOTGate(), [0, 3])
circuit.append_gate(TGate(), [3])
circuit.append_gate(CNOTGate(), [0, 3])

circuit.append_gate(LogicalAndDgGate(), [1, 2, 3])

circuit.append_gate(CNOTGate(), [1, 2])
circuit.append_gate(CNOTGate(), [0, 1])
circuit.append_gate(CNOTGate(), [0, 2])

circuit.append_gate(HGate(), [0])
circuit.append_gate(HGate(), [1])
circuit.append_gate(XGate(), [2])
circuit.append_gate(HGate(), [2])

```

```python
from bqskit.qis import StateVector

input_state = StateVector.zero(4)

out_state = circuit.get_statevector(input_state)

all_probs = out_state.get_probs()
total_qubits = 4
num_qubits = 3

for qubit_ind in range(num_qubits):
    prob = sum(prob for i, prob in enumerate(all_probs) if (i & (1 << (total_qubits - 1 - qubit_ind))) != 0)
    print(f"Probability of qubit {qubit_ind} being 1: {prob:.4f}")
```

```python
# Try without qubit 1
circuit = Circuit(3)

circuit.append_gate(HGate(), [0])

circuit.append_gate(HGate(), [1])
circuit.append_gate(SqrtTGate(), [1])

# Should kick back sqrt(T) on qubit 0
circuit.append_gate(HGate(), [0])


# Requires a T state input on qubit 2 for LogicalAndGate
circuit.append_gate(HGate(), [2])
circuit.append_gate(TGate(), [2])

# Now run circuit
circuit.append_gate(XGate(), [2])
circuit.append_gate(CNOTGate(), [0, 1])

circuit.append_gate(LogicalAndGate(), [0, 1, 2])

circuit.append_gate(CNOTGate(), [0, 2])
circuit.append_gate(TGate(), [2])
circuit.append_gate(CNOTGate(), [0, 2])

circuit.append_gate(LogicalAndDgGate(), [0, 1, 2])

circuit.append_gate(CNOTGate(), [0, 1])

circuit.append_gate(HGate(), [0])
circuit.append_gate(XGate(), [1])
circuit.append_gate(HGate(), [1])
```

```python
from bqskit.qis import StateVector

input_state = StateVector.zero(3)

out_state = circuit.get_statevector(input_state)

all_probs = out_state.get_probs()
total_qubits = 3
num_qubits = 2

for qubit_ind in range(num_qubits):
    prob = sum(prob for i, prob in enumerate(all_probs) if (i & (1 << (total_qubits - 1 - qubit_ind))) != 0)
    print(f"Probability of qubit {qubit_ind} being 1: {prob:.4f}")
```
